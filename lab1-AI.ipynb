```python
import nltk
import spacy
import string
import re
from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize
from nltk.stem import WordNetLemmatizer

# Download NLTK resources
nltk.download('punkt')
nltk.download('punkt_tab')      # needed for newer NLTK versions
nltk.download('stopwords')
nltk.download('wordnet')
nltk.download('omw-1.4')

# Load spaCy model
nlp = spacy.load("en_core_web_sm")
```

```python
#Partie 2 — Nettoyage du texte

text = "Hello,Everyone!!! are interseted to join our univers ,I'm not ... the year @2024 was not it  #Python is a cool language 100% fun :)"

# 1. Lowercase
text_lower = text.lower()
print("Avant :", text)
print("Après minuscules :", text_lower)

# 2. Remove punctuation (with str.translate)
text_clean = text_lower.translate(str.maketrans('', '', string.punctuation))
print("Après nettoyage :", text_clean)

# Alternative with regex
text_clean_re = re.sub(r'[^\w\s]', '', text_lower)
print("Avec regex :", text_clean_re)
```

```python
# Partie 3 — Tokenisation et Stopwords
text = "Hello,Everyone!!! are interseted to join our univers ,I'm not ... the year @2024 was not it  #Python is a cool language 100% fun :)"
text_lower = text.lower()
text_clean = re.sub(r'[^\w\s]', '', text_lower)

# Tokenization with NLTK
tokens = word_tokenize(text_clean)
print("Tokens :", tokens)
# ['hello', 'world', 'nlp', 'is', 'amazing', 'isnt', 'it', 'python', '2024', '100', 'fun']

# Stopwords (English by default)
stop_words = set(stopwords.words('english'))
print("Nombre de stopwords :", len(stop_words))
print("Exemples :", list(stop_words)[:20])

# Filter stopwords
filtered_tokens = [w for w in tokens if w not in stop_words]
print("Tokens filtrés :", filtered_tokens)
# ['hello', 'world', 'nlp', 'amazing', 'isnt', 'python', '2024', '100', 'fun']

# Add a custom stopword (e.g., "isnt" adds no meaning here)
stop_words.add("isnt")
filtered_tokens = [w for w in tokens if w not in stop_words]
print("Après ajout de 'isnt' :", filtered_tokens)
```

```python
removed = len(tokens) - len(filtered_tokens)
ratio = removed / len(tokens) * 100
print(f"{removed} mots supprimés sur {len(tokens)} → {ratio:.1f}%")
```

```python
#Partie 4 — Lemmatisation
lemmatizer = WordNetLemmatizer()

# Before lemmatization
print("Avant :", filtered_tokens)

# Lemmatize each token (default POS = noun)
lemmatized = [lemmatizer.lemmatize(w) for w in filtered_tokens]
print("Après (noun) :", lemmatized)

# Better: with POS tagging via spaCy or NLTK
lemmatized_pos = [lemmatizer.lemmatize(w, pos='v') for w in filtered_tokens]
print("Après (verbe) :", lemmatized_pos)

# Comparison table
for before, after in zip(filtered_tokens, lemmatized):
    if before != after:
        print(f"{before:15} → {after}")
```

```python
sample = ["running", "runs", "ran", "better", "best", "dogs", "children"]
print([lemmatizer.lemmatize(w) for w in sample])
# ['running', 'runs', 'ran', 'better', 'best', 'dog', 'child']
# Note: 'running'/'runs'/'ran' need POS='v' to be reduced to 'run'
print([lemmatizer.lemmatize(w, pos='v') for w in sample])
# ['run', 'run', 'run', 'better', 'best', 'dogs', 'children']
```
